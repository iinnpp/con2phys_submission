### Q12 visualization: normalized within-trial firing rates for behavioral variable C

The same trial-selection criteria as Q11 are applied directly to the source trial table: all four event times must be present and strictly ordered, and the final recorded trial for each mouse is excluded. No filtering or balancing is applied to variable C.

Trials have different durations, so every retained trial is warped piecewise into three normalized phases:

1. `trial_start` to `stim_start`
2. `stim_start` to `outcome`
3. `outcome` to `trial_end`

Each phase receives the same number of bins. This aligns trial start, stimulus, outcome, and trial end across all trials. Firing rates are calculated using each bin's true duration in seconds, so the y-axis remains in Hz. The normalized x-axis represents trial phase rather than physical time.

Within each mouse and brain area, firing rates are first averaged equally across recorded units for each trial and then averaged across trials. One figure is saved per mouse. Each figure contains one panel per brain area, with separate colored trajectories for the observed values C=0, C=1, and C=2. Shaded bands show the trial-level SEM when at least two trials are available; a singleton C level is plotted without an SEM band.

```python
import csv
import os
from collections import defaultdict
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

#### Parameters

```python
project_folder = Path(os.environ.get(
    "CON2PHYS_PROJECT_DIR", "E:/con2phys_submission"
))
repository_folder = Path(os.environ.get(
    "CON2PHYS_REPOSITORY_DIR", "D:/git_repos/con2phys_submission"
))
source_trials_csv_path = Path(os.environ.get(
    "CON2PHYS_TRIALS_CSV",
    project_folder / "dataset" / "trials_info.csv",
))
sua_csv_path = Path(os.environ.get(
    "CON2PHYS_SUA_CSV",
    project_folder / "dataset" / "full_sua_dataset.csv",
))
q_path = repository_folder / "results" / "q12"
trials_csv_path = q_path / "trials_info_filtered.csv"
per_mouse_plot_folder = q_path / "per_mouse_plots"

EXPECTED_N_MICE = 18
EXPECTED_SOURCE_TRIALS = 10_373
EXPECTED_FILTERED_TRIALS = 10_350
BRAIN_AREAS = (1, 2, 3)
C_LEVELS = (0, 1, 2)
N_BINS_PER_PHASE = 20
N_PHASES = 3
N_NORMALIZED_BINS = N_BINS_PER_PHASE * N_PHASES

rates_csv_path = q_path / "all_mice_normalized_trial_firing_rates.csv"
```

#### Apply the Q11 trial-selection criteria

The final recorded trial is defined from the unfiltered source table as the maximum `trial_id` for each mouse. The filtered CSV is saved in the Q12 results directory before the firing-rate analysis begins.

```python
if not source_trials_csv_path.exists():
    raise FileNotFoundError(f"Source trial table not found: {source_trials_csv_path}")

source_trials_df = pd.read_csv(source_trials_csv_path)
expected_trial_columns = [
    "mouse_id", "trial_id", "variable_A", "variable_B", "variable_C",
    "trial_start", "stim_start", "outcome", "trial_end", "trial_duration",
]
if list(source_trials_df.columns) != expected_trial_columns:
    raise ValueError(f"Unexpected trial schema: {list(source_trials_df.columns)}")
if source_trials_df[["mouse_id", "trial_id"]].duplicated().any():
    raise ValueError("Duplicate mouse/trial identifiers found in source table")
if len(source_trials_df) != EXPECTED_SOURCE_TRIALS:
    raise ValueError(
        f"Expected {EXPECTED_SOURCE_TRIALS:,} source trials, "
        f"found {len(source_trials_df):,}"
    )

timing_columns = ["trial_start", "stim_start", "outcome", "trial_end"]
complete_timing = source_trials_df[timing_columns].notna().all(axis=1)
correct_order = (
    source_trials_df["trial_start"].lt(source_trials_df["stim_start"])
    & source_trials_df["stim_start"].lt(source_trials_df["outcome"])
    & source_trials_df["outcome"].lt(source_trials_df["trial_end"])
)
source_final_trial_id = source_trials_df.groupby("mouse_id")["trial_id"].transform("max")
is_final_recorded_trial = source_trials_df["trial_id"].eq(source_final_trial_id)
retain_trial = complete_timing & correct_order & ~is_final_recorded_trial

trials_df = (
    source_trials_df.loc[retain_trial, expected_trial_columns]
    .sort_values(["mouse_id", "trial_id"])
    .reset_index(drop=True)
)
if len(trials_df) != EXPECTED_FILTERED_TRIALS:
    raise ValueError(
        f"Expected {EXPECTED_FILTERED_TRIALS:,} retained trials, "
        f"found {len(trials_df):,}"
    )
if trials_df[expected_trial_columns].isna().any().any():
    raise ValueError("Filtered trial table contains missing values")
if not (
    trials_df["trial_start"].lt(trials_df["stim_start"])
    & trials_df["stim_start"].lt(trials_df["outcome"])
    & trials_df["outcome"].lt(trials_df["trial_end"])
).all():
    raise ValueError("Filtered trial table contains incorrectly ordered events")
if (
    trials_df["trial_id"]
    == trials_df["mouse_id"].map(
        source_trials_df.groupby("mouse_id")["trial_id"].max()
    )
).any():
    raise ValueError("A final recorded trial remains in the filtered table")

calculated_duration = trials_df["trial_end"] - trials_df["trial_start"]
if not np.allclose(
    trials_df["trial_duration"], calculated_duration, rtol=0, atol=1e-9
):
    raise ValueError("trial_duration does not match trial_end - trial_start")
if set(trials_df["variable_C"].unique()) != set(C_LEVELS):
    raise ValueError(
        f"variable_C must contain {C_LEVELS}; "
        f"found {sorted(trials_df['variable_C'].unique())}"
    )

mouse_ids = sorted(int(mouse_id) for mouse_id in trials_df["mouse_id"].unique())
if len(mouse_ids) != EXPECTED_N_MICE:
    raise ValueError(f"Expected {EXPECTED_N_MICE} mice, found {len(mouse_ids)}")

q_path.mkdir(parents=True, exist_ok=True)
trials_df.to_csv(trials_csv_path, index=False)
trials_by_mouse = {
    int(mouse_id): group.sort_values("trial_id").reset_index(drop=True)
    for mouse_id, group in trials_df.groupby("mouse_id", sort=True)
}
class_counts = (
    pd.crosstab(trials_df["mouse_id"], trials_df["variable_C"])
    .reindex(index=mouse_ids, columns=C_LEVELS, fill_value=0)
)

print(
    f"Timing-complete, correctly ordered trials: "
    f"{int((complete_timing & correct_order).sum()):,}"
)
print(
    f"Excluded final trials that otherwise passed timing criteria: "
    f"{int((complete_timing & correct_order & is_final_recorded_trial).sum())}"
)
print(f"Saved {len(trials_df):,} filtered trials to {trials_csv_path}")
display(class_counts.rename(columns={level: f"C={level}" for level in C_LEVELS}))
```

#### Construct trial-specific normalized bins

The same normalized bin has different physical widths in different trials. `bin_duration_sec` stores those true widths and is used when converting spike counts to Hz. No spikes from inter-trial intervals are included.

```python
def piecewise_trial_edges(trial_row, n_bins_per_phase):
    event_times = np.array([
        trial_row["trial_start"],
        trial_row["stim_start"],
        trial_row["outcome"],
        trial_row["trial_end"],
    ], dtype=float)
    phase_edges = []
    for phase_index in range(3):
        edges = np.linspace(
            event_times[phase_index],
            event_times[phase_index + 1],
            n_bins_per_phase + 1,
        )
        if phase_index:
            edges = edges[1:]
        phase_edges.append(edges)
    return np.concatenate(phase_edges)


trial_edges_by_mouse = {}
bin_durations_by_mouse = {}
for mouse_id, mouse_trials in trials_by_mouse.items():
    edges = np.vstack([
        piecewise_trial_edges(row, N_BINS_PER_PHASE)
        for _, row in mouse_trials.iterrows()
    ])
    durations = np.diff(edges, axis=1)
    expected_shape = (len(mouse_trials), N_NORMALIZED_BINS + 1)
    if edges.shape != expected_shape:
        raise ValueError(f"Unexpected normalized edge shape for mouse {mouse_id}")
    if not np.isfinite(durations).all() or np.any(durations <= 0):
        raise ValueError(f"Invalid normalized durations for mouse {mouse_id}")
    trial_edges_by_mouse[mouse_id] = edges
    bin_durations_by_mouse[mouse_id] = durations

normalized_centers = (
    np.arange(N_NORMALIZED_BINS, dtype=float) + 0.5
) / N_BINS_PER_PHASE
phase_names = np.repeat([
    "trial_start_to_stim", "stim_to_outcome", "outcome_to_trial_end"
], N_BINS_PER_PHASE)
phase_fractions = np.tile(
    (np.arange(N_BINS_PER_PHASE) + 0.5) / N_BINS_PER_PHASE,
    N_PHASES,
)

portion_duration_summary = (
    trials_df.assign(
        start_to_stim=trials_df["stim_start"] - trials_df["trial_start"],
        stim_to_outcome=trials_df["outcome"] - trials_df["stim_start"],
        outcome_to_end=trials_df["trial_end"] - trials_df["outcome"],
    )
    .groupby("mouse_id")[[
        "start_to_stim", "stim_to_outcome", "outcome_to_end"
    ]]
    .median()
)
display(portion_duration_summary)
```

#### Calculate population firing rates

For every unit and trial, spikes are counted in the trial-specific normalized bins. Each count is divided by its true bin duration. Unit rates are then averaged within brain area, giving one population-mean firing-rate trajectory per trial and area.

```python
expected_sua_columns = [
    "mouse_id", "cluster_id", "brain_area", "recording_duration_sec",
    "unit_last_spike_sec", "n_spikes", "spikes",
]
area_rate_sums = {
    (mouse_id, brain_area): np.zeros(
        (len(trials_by_mouse[mouse_id]), N_NORMALIZED_BINS), dtype=float
    )
    for mouse_id in mouse_ids
    for brain_area in BRAIN_AREAS
}
unit_counts = defaultdict(int)
unit_keys = set()

csv.field_size_limit(2**31 - 1)
with sua_csv_path.open(newline="", encoding="utf-8") as handle:
    reader = csv.DictReader(handle)
    if reader.fieldnames != expected_sua_columns:
        raise ValueError(f"Unexpected SUA schema: {reader.fieldnames}")

    for row in reader:
        mouse_id = int(row["mouse_id"])
        if mouse_id not in trials_by_mouse:
            continue

        cluster_id = int(row["cluster_id"])
        brain_area = int(row["brain_area"])
        unit_key = (mouse_id, cluster_id)
        group_key = (mouse_id, brain_area)
        if unit_key in unit_keys:
            raise ValueError(f"Duplicate unit key: {unit_key}")
        if brain_area not in BRAIN_AREAS:
            raise ValueError(f"Unexpected brain area for {unit_key}: {brain_area}")

        spike_times = np.fromstring(row["spikes"], sep=";", dtype=np.float64)
        expected_n_spikes = int(row["n_spikes"])
        if spike_times.size != expected_n_spikes or spike_times.size == 0:
            raise ValueError(f"Spike-count mismatch for {unit_key}")
        if not np.isfinite(spike_times).all() or np.any(np.diff(spike_times) < 0):
            raise ValueError(f"Invalid or unsorted spikes for {unit_key}")

        edges = trial_edges_by_mouse[mouse_id]
        durations = bin_durations_by_mouse[mouse_id]
        edge_indices = np.searchsorted(
            spike_times, edges.reshape(-1), side="left"
        ).reshape(edges.shape)
        unit_rates = np.diff(edge_indices, axis=1) / durations

        area_rate_sums[group_key] += unit_rates
        unit_counts[group_key] += 1
        unit_keys.add(unit_key)

expected_group_keys = {
    (mouse_id, brain_area)
    for mouse_id in mouse_ids
    for brain_area in BRAIN_AREAS
}
if set(unit_counts) != expected_group_keys:
    raise ValueError(
        f"Missing mouse/area groups: {sorted(expected_group_keys - set(unit_counts))}"
    )

population_rates = {
    group_key: area_rate_sums[group_key] / unit_counts[group_key]
    for group_key in expected_group_keys
}
if not all(np.isfinite(values).all() for values in population_rates.values()):
    raise ValueError("Population firing rates contain non-finite values")

unit_counts_df = pd.DataFrame([
    {
        "mouse_id": mouse_id,
        "brain_area": brain_area,
        "n_units": unit_counts[(mouse_id, brain_area)],
    }
    for mouse_id, brain_area in sorted(expected_group_keys)
])
display(unit_counts_df)
```

#### Summarize and compare C=0, C=1, and C=2

Only C levels observed for a given mouse are drawn in that mouse's figure. Some mice have no C=1 trials, and one mouse has a single C=1 trial; the singleton mean is retained but its SEM is undefined and therefore not shaded.

```python
summary_rows = []
class_colors = {0: "tab:blue", 1: "tab:orange", 2: "tab:green"}
event_positions = [0, 1, 2, 3]
event_labels = ["Trial start", "Stimulus", "Outcome", "Trial end"]
per_mouse_plot_folder.mkdir(parents=True, exist_ok=True)

for mouse_id in mouse_ids:
    mouse_trials = trials_by_mouse[mouse_id]
    labels = mouse_trials["variable_C"].to_numpy(dtype=int)
    observed_levels = [level for level in C_LEVELS if np.any(labels == level)]
    class_summaries = {}

    for variable_c in observed_levels:
        class_mask = labels == variable_c
        n_class_trials = int(class_mask.sum())
        class_summaries[variable_c] = {}
        for brain_area in BRAIN_AREAS:
            group_key = (mouse_id, brain_area)
            trial_rates = population_rates[group_key][class_mask]
            mean_rate = trial_rates.mean(axis=0)
            if n_class_trials >= 2:
                sem_rate = (
                    trial_rates.std(axis=0, ddof=1) / np.sqrt(n_class_trials)
                )
            else:
                sem_rate = np.full(N_NORMALIZED_BINS, np.nan)
            class_summaries[variable_c][brain_area] = (mean_rate, sem_rate)

            for bin_index in range(N_NORMALIZED_BINS):
                summary_rows.append({
                    "mouse_id": mouse_id,
                    "brain_area": brain_area,
                    "variable_C": variable_c,
                    "normalized_time": normalized_centers[bin_index],
                    "phase": phase_names[bin_index],
                    "phase_fraction": phase_fractions[bin_index],
                    "mean_firing_rate_hz": mean_rate[bin_index],
                    "sem_across_trials_hz": sem_rate[bin_index],
                    "n_trials": n_class_trials,
                    "n_units": unit_counts[group_key],
                })

    upper_values = []
    for class_data in class_summaries.values():
        for mean_rate, sem_rate in class_data.values():
            finite_sem = np.where(np.isfinite(sem_rate), sem_rate, 0.0)
            upper_values.append(mean_rate + finite_sem)
    common_y_max = max(values.max() for values in upper_values) * 1.05

    fig, axes = plt.subplots(
        1, len(BRAIN_AREAS), figsize=(13, 3.8), sharex=True, sharey=True
    )
    for axis, brain_area in zip(axes, BRAIN_AREAS):
        for variable_c in observed_levels:
            mean_rate, sem_rate = class_summaries[variable_c][brain_area]
            color = class_colors[variable_c]
            n_class_trials = int((labels == variable_c).sum())
            axis.plot(
                normalized_centers, mean_rate, color=color, linewidth=1.8,
                label=f"C={variable_c} (n={n_class_trials})",
            )
            if n_class_trials >= 2:
                axis.fill_between(
                    normalized_centers,
                    mean_rate - sem_rate,
                    mean_rate + sem_rate,
                    color=color,
                    alpha=0.22,
                    linewidth=0,
                )
        for event_position in event_positions[1:-1]:
            axis.axvline(
                event_position, color="0.35", linestyle="--", linewidth=0.9
            )
        axis.set_title(
            f"Brain area {brain_area} "
            f"(n={unit_counts[(mouse_id, brain_area)]} units)"
        )
        axis.set_xticks(event_positions, event_labels, rotation=20)
        axis.set_xlim(0, 3)
        axis.set_ylim(0, common_y_max)
        axis.set_xlabel("Normalized trial phase")
        axis.grid(True, axis="y", alpha=0.2)
        axis.legend(frameon=False, loc="best")
        sns.despine(ax=axis)

    axes[0].set_ylabel("Population mean firing rate (Hz)")
    fig.suptitle(
        f"Mouse {mouse_id}: normalized within-trial firing rate by variable C"
    )
    fig.tight_layout()
    figure_path = (
        per_mouse_plot_folder
        / f"mouse_{mouse_id:02d}_normalized_firing_rate_by_C.png"
    )
    fig.savefig(figure_path, dpi=200, bbox_inches="tight")
    plt.close(fig)
    print(f"Saved mouse {mouse_id} figure: {figure_path.name}")

normalized_rates_df = pd.DataFrame(summary_rows)
normalized_rates_df.to_csv(rates_csv_path, index=False)

display(normalized_rates_df.head())
print(f"Saved {len(normalized_rates_df):,} normalized summary rows.")
print(f"Saved normalized data to {rates_csv_path}")
print(f"Saved {len(mouse_ids)} figures to {per_mouse_plot_folder}")
```
